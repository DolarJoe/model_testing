## Testing of connectivity

The idea behind testing connectivity handling of a simulator is relatively straightforward. Suppose that there are 2 equivalent models written in C, except one of them uses 32bit floats to represent their floating point numbers, and the other uses 64bit doubles. Given enough steps, these models, although equivalent otherwise, will diverge because of rounding errors, giving us a false negative on the models being equivalent. 

To avoid this issue, we've come up with the solution of running only 1 step at a time, as if those results are close enough, we know the reasoning behind the models is equivalent.
